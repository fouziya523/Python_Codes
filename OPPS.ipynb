```python
# ============================================================
# SIMPLE LOGIN / PASSWORD PROGRAM
# Main idea:
# Dictionary stores:
# USER_ID  --->  [NAME, PASSWORD]
#
# Example:
# "123456" ---> ["fouziya", "111"]
# ============================================================


class login:

    # __init__ runs automatically when we create an object.
    # It receives the user's name, user ID and password.
    def __init__(self, name, userid, password):

        # Create an empty dictionary.
        # This dictionary will store the user's information.
        self.d = {}

        # Store the user's name inside the object.
        self.name = name

        # Store the user's ID inside the object.
        self.userid = userid

        # Store the password.
        # __password means this is intended to be a private attribute.
        self.__password = password

        # Store the user's information in the dictionary.
        #
        # self.userid becomes the KEY.
        # [self.name, self.__password] becomes the VALUE.
        #
        # So it looks like:
        # {
        #     "123456": ["fouziya", "111"]
        # }
        self.d[self.userid] = [self.name, self.__password]


    # This function is called when the user forgets their password.
    def forgot_password(self):

        # Ask the user to enter their user ID.
        uid = input("Provide your user ID: ")

        # Check whether the entered ID exists in our dictionary.
        #
        # "uid in self.d" means:
        # "Does this user ID exist as a KEY in the dictionary?"
        if uid in self.d:

            # self.d[uid] gives the user's information.
            #
            # Example:
            # self.d[uid]
            # gives:
            # ["fouziya", "111"]
            #
            # [1] means the SECOND item.
            # [0] = name
            # [1] = password
            return self.d[uid][1]

        # If the user ID does not exist.
        else:
            return "User ID not found."


    # This function changes the password.
    # 'new' is the new password given by the user.
    def change_password(self, new):

        # Change the private password variable.
        self.__password = new

        # Update the dictionary with the new password.
        #
        # Example:
        # OLD:
        # "123456": ["fouziya", "111"]
        #
        # NEW:
        # "123456": ["fouziya", "new123"]
        self.d[self.userid] = [self.name, self.__password] #same hi hai but new pw assinfment kare


# ============================================================
# CREATE A USER
# ============================================================

# Create an object called x.
#
# name     = fouziya
# userid   = 123456
# password = 111
x = login( # function call k jaisa hai yee
    name="fouziya",
    userid="123456",
    password="111"
)


# ============================================================
# FORGOT PASSWORD
# ============================================================

# Call the forgot_password() function.
#
# It asks:
# "Provide your user ID:"
#
# If we enter 123456,
# it finds the user and returns the password.
print(x.forgot_password())


# ============================================================
# CHANGE PASSWORD
# ============================================================

# Change the password from:
# "111"
#
# to:
# "new123"
x.change_password("new123")


# ============================================================
# CHECK PASSWORD AGAIN
# ============================================================

# Call forgot_password() again.
#
# Now the password should be:
# "new123"
print(x.forgot_password())
```

```python
#abstraction : methods , used to design codes
from abc import ABC,abstractmethod
class A(ABC):
  @abstractmethod
  def m1(self):
    return 'm1'
  def m2(self):
    return 'm2'
  def m3(self):
    return 'm3'
class subclass(A): #+polymorphimium
  def m1(self):
    return 'this is m1 from subclass'
  def m2(self):
    return 'this is m2 from subclass'


# d=A()
# d.m1()
c=subclass()
print(c.m1())
c.m2()
```

```python
from abc import ABC, abstractmethod

# ABC = tells Python this is an Abstract Base Class
class A(ABC):

    # abstractmethod = child classes MUST create their own m1()
    @abstractmethod
    def m1(self):
        pass

    # Normal method
    # Child gets this automatically if it doesn't override it
    def m2(self):
        return 'm2'

    # Normal method
    def m3(self):
        return 'm3'


# Child class
class subclass(A):

    # We MUST define m1(), because parent marked it abstract
    def m1(self):
        return 'this is m1 from subclass'

    # We are overriding parent's m2()
    def m2(self):
        return 'this is m2 from subclass'


# Now we can create an object of subclass
c = subclass()

print(c.m1())
# Output:
# this is m1 from subclass

print(c.m2())
# Output:
# this is m2 from subclass

print(c.m3())
# Output:
# m3
```

```python
from abc import ABC,abstractmethod
class PaymentProcessor(ABC):
  @abstractmethod
  def payment(self):
    pass # Abstract methods don't have an implementation in the abstract class

class CashPayment(PaymentProcessor): # A concrete subclass
  def payment(self):
    return 'Payment made by cash'

class CardPayment(PaymentProcessor):
  def payment(self):
    return 'Payment made by card'

# Now, instantiate the concrete subclass, not the abstract base class
cash_processor = CashPayment()
print(cash_processor.payment())

card_processor = CardPayment()
print(card_processor.payment())
```

```python
# =====================================================
# ABSTRACTION - EASY TEMPLATE
# =====================================================

from abc import ABC, abstractmethod


# 1. Create ABSTRACT PARENT class
class Parent(ABC):

    # 2. Create ABSTRACT METHOD
    #    Child MUST implement this method
    @abstractmethod
    def method(self):
        pass


# 3. Create CHILD class
class Child1(Parent):

    # 4. Implement the abstract method
    def method(self):
        return "Child 1 implementation"


# 5. Another CHILD class
class Child2(Parent):

    # 6. Same method, different implementation
    def method(self):
        return "Child 2 implementation"


# 7. Create objects of CHILD classes
obj1 = Child1()
obj2 = Child2()


# 8. Call the same method
print(obj1.method())
print(obj2.method())


# =====================================================
# REMEMBER:
#
# ABC = Abstract Base Class
#
# @abstractmethod =
# "Child MUST implement this method"
#
# Parent = RULE
# Child = IMPLEMENTATION
#
# Same method + different behavior
# = POLYMORPHISM
#
# You cannot directly create an object
# of a class having unimplemented abstract methods.
# =====================================================
```

```python
from abc import ABC, abstractmethod

# Parent / Abstract class
class payment(ABC):

    # Child MUST implement upi()
    @abstractmethod
    def upi(self):
        pass

    # Normal methods
    def banktoback(self):
        pass

    def phonepay(self):
        pass


# Child class
class subclass(payment):

    # Implementing the abstract method
    def upi(self):
        return 'this is upi payment'

    # Overriding parent method
    def banktoback(self):
        return 'this is bank to bank payment'

    # Overriding parent method
    def phonepay(self):
        return 'this is phonepay payment'


# Create object
x = subclass()

# CALL the methods
print(x.upi())
print(x.banktoback())
print(x.phonepay())
```
