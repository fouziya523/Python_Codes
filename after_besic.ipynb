```python
#map function:map function maps a function with a interable

def fun1(x):
  if x==2:
    return 'yes'
  else:
    return 'no'
m=[12,3,4,2,11]
list(map(fun1,m))
```

```python
#filter: no use of loops, like maps
def fun1(x):
  if x==2:
    return True
  else:
    return False
c=[12,3,2,12,2,4,2]
list(filter(fun1,c))
```
