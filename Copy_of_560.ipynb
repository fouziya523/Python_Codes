```python
# set built in function
a={1,2,3,90,"kl"}
a.add("fouziya") # adds one element at a time
print("After add('fouziya'):", a)
#update
# The update() method itself returns None, so printing its result will always show None.
# The actual set 'a' is modified by the update operation.
_ = a.update("fouziya") # "letters ku khulla likhra" ( adds multiple elements at a time)
print("After update('fouziya'):", a)

_ = a.update() # update with no arguments does nothing
print("After update():", a)
#pop removes an arbitrary element from the set
popped_item_1 = a.pop()
print("Popped item 1:", popped_item_1)
print("Set after first pop:", a)

popped_item_2 = a.pop() # remove values from the set
print("Popped item 2:", popped_item_2)
print("Set after second pop:", a)

# Re-initialize 'a' to demonstrate remove and discard without KeyError after clear()
a={1,2,3,90,"kl"}
print("Re-initialized set 'a':", a)

# discard(element) removes an element if present, does nothing if not present
# It returns None.
_ = a.discard(12) # 12 is not in the set, so nothing happens
print("After discard(12):", a)

# remove(element) removes an element if present, raises KeyError if not present
# It returns None.
_ = a.remove(1)
print("After remove(1):", a)

# sorted() raises TypeError when trying to compare str and int.
# To sort, either filter out non-comparable types or convert all to a common type (e.g., strings).
print("Sorted elements of set 'a' (converted to strings):", sorted([str(x) for x in a]))
```
