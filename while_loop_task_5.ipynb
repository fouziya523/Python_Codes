```python
# Print numbers from 1 to 10 using a while loop.
i=1
while 10>=i:
  print(i)
  i+=1
```

```python
# Print numbers from 10 to 1 using a while loop.
i=10
while i<=10:
  print(i)
  i-=1
  if i==0:
    break
```
